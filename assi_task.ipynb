```python
text = """Haruki Murakami is a Japanese writer. His novels, essays, and short stories have been best-sellers in Japan and internationally, with his work translated into 50 language and having sold millions of copies outside Japan.He has received numerous awards for his work, including the Gunzo Prize for New Writers, the World Fantasy Award, the Tanizaki Prize, Yomiuri Prize for Literature, the Frank O'Connor International Short Story Award, the Noma Literary Prize, the Franz Kafka Prize, the Kiriyama Prize for Fiction, the Goodreads Choice Awards for Best Fiction, the Jerusalem Prize, and the Princess of Asturias Awards.
Murakami was born in Kyoto, Japan, during the post-World War II baby boom and was raised in Nishinomiya, Ashiya and Kobe.He is an only child. His father was the son of a Buddhist pri
est, and his mother is the daughter of an Osaka merchant.Both taught Japanese literature.His father was involved in the Second Sino-Japanese War, and was deeply traumatized by it, which would, in turn, affect Murakami.
Murakami was heavily influenced in childhood by Western culture, particularly Russian music and literature. He grew up reading a wide range of works by European and American writers, such as Franz Kafka, Gustave Flaubert, Charles Dickens, Kurt Vonnegut, Fyodor Dostoyevsky, Richard Brautigan and Jack Kerouac.These Western influences distinguish Murakami from the majority of other Japanese writers."""
```

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# Lowercasing
text = text.lower()
```

```python
# Tokenization
from nltk.tokenize import word_tokenize
lower_tokens = word_tokenize(text)
print(lower_tokens)
```

```python
# stopword removal
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))    # return set of english stopWords

filterd_tokens = [
    word for word in lower_tokens    # using list comprehension
    if word.isalpha() and word not in stop_words   # isalpha removes punctuation, numbers, symbols
]

print(filterd_tokens)
```

```python
# Stemming
# Cuts words to their root form

from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

print(f"before stemming: {filterd_tokens}")
stemmed_words = [stemmer.stem(word) for word in filterd_tokens]
print(f"after stemming: {stemmed_words}")
```

```python
# Lemmatization
# Converts words to dictionary form
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filterd_tokens]
print(lemmatized_words)
```

### `Task 5` : Build a Mini Text Analyzer
<pre>
Goal: Bring everything together.
What to Do:
1. Collect 10 short sentences or tweets.
2. Preprocess them (tokenize, clean, remove stopwords).
3. Find most common words and top 5 bigrams.
4. Print results neatly.
Concept Focus: Apply all text cleaning techniques to a small corpus.
</pre>


```python
# Load dataset
import pandas as pd 
df = pd.read_csv("irony_detection_assignment/SemEval2018-T3_gold_test_taskA_emoji.txt",sep = "\t")
tweets = df["Tweet text"]
print(tweets)
```

```python
# converting to list
tweet_list = tweets.tolist()
print(tweet_list[:5])
```

```python
from nltk.tokenize import word_tokenize 
from nltk.corpus import stopwords 
import nltk 
# nltk.download('punkt')
# nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

cleaned_tweets = []

for tweet in tweet_list:
    tokens = word_tokenize(tweet.lower())
    tokens = [word for words in tokens if word.isalpha() and word not in stop_words]
    cleaned_tweets.append(tokens)
    
print(cleaned_tweets[:5])
```

```python
from collections import Counter
from nltk.util import ngrams
```

```python
all_tokens = [word for tweet in cleaned_tweets for word in tweet]
```

```python
word_freq = Counter(all_tokens)

print("top 10 Most Common words:")
for word, count in word_freq.most_common(20):
    print(f"{word:<15}{count}")
```

```python
all_bigrams = list(ngrams(all_tokens,2))
```

```python
bigram_freq = Counter(all_bigrams)
```

```python
print("top 10 bigrams")
for bigram, count in bigram_freq.most_common(10):
    print(f"{bigram} -> {count}")
```

### `Task 6` : Bag of Words and TF-IDF
<pre>
Goal: Convert text into numbers.
What to Do:
1. Take 4–5 small documents.
2. Use CountVectorizer and TfidfVectorizer from scikit-learn.
3. Print BoW and TF-IDF matrices.
4. Compare what changed.
Concept Focus: Move from raw text → numeric form.
</pre>


```python
documents = [
    "The famous book series 'Harry Potter' is written by J.K. Rowling",
    "Karkroff was a death eater in the Harry Potter series",
    "The death mark in 'The goblet of Fire' was created by the Harry's wand but not by him",
    "Winky used to work at the Hogwarts school",
    "Dobby was a free elf who helped Harry Potter"
]

# import libraries
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer

import pandas as pd
```

```python
count_vectorizer = CountVectorizer(stop_words='english')

bow_matrix = count_vectorizer.fit_transform(documents)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

print("bag of words matrix : \n")
print(bow_df)
```

```python
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)
print(f"\nTF-IDF Matrix : \n{tfidf_df.round(2)}")
```

### `Task 7` : Document Similarity
<pre>
Goal: Measure how similar two texts are.
What to Do:
1. Use the TF-IDF matrix from Task 6.
2. Compute pairwise cosine similarity.

3. Print the similarity matrix.
4. Discuss which documents are similar and why.
Concept Focus: Understanding cosine similarity as a distance measure.
</pre>


```python
from sklearn.metrics.pairwise import cosine_similarity


similarity_matrix = cosine_similarity(tfidf_matrix)

similarity_df = pd.DataFrame(
    similarity_matrix,
    columns=[f"Doc {i}" for i in range(1,6)],
    index = [f"Doc {i}" for i in range(1,6)]
) 

print("Cosine Similarity Matrix :\n")
print(similarity_df.round(2))
```
